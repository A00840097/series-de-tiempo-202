```python
import pandas as pd
import numpy as np
from statsmodels.tsa.stattools import adfuller, arma_order_select_ic
from statsmodels.tsa.api import ARIMA
from statsmodels.tsa.vector_ar.vecm import coint_johansen
import os
import warnings

warnings.filterwarnings("ignore")

# Verify uploaded files exist
apple_filename = "/content/APPLE 1 MIN.xlsx"
msft_filename = "/content/MICROSOFT 1 MIN.xlsx"

if not os.path.exists(apple_filename):
    raise FileNotFoundError(f"Error: '{apple_filename}' not found. Upload manually in the file explorer.")

if not os.path.exists(msft_filename):
    raise FileNotFoundError(f"Error: '{msft_filename}' not found. Upload manually in the file explorer.")

# Load data
apple_df = pd.read_excel(apple_filename)
msft_df = pd.read_excel(msft_filename)

# Extract closing prices
try:
    apple_close = apple_df['Close']
    msft_close = msft_df['Close']
except KeyError as e:
    print("Error: 'Close' column not found.")
    print("Apple columns:", apple_df.columns.tolist())
    print("Microsoft columns:", msft_df.columns.tolist())
    raise e

# Align series lengths
min_length = min(len(apple_close), len(msft_close))
apple_close = apple_close[:min_length]
msft_close = msft_close[:min_length]

# ADF unit root test
def adf_test(series, name):
    result = adfuller(series, autolag='AIC')
    print(f'\nADF Test for {name}:')
    print(f'ADF Statistic: {result[0]:.4f}')
    print(f'p-value: {result[1]:.4f}')
    print('Stationary' if result[1] < 0.05 else 'Non-stationary')
    return result[1] < 0.05

apple_stationary = adf_test(apple_close, 'Apple Close')
msft_stationary = adf_test(msft_close, 'Microsoft Close')

# Difference series if non-stationary
apple_series = apple_close if apple_stationary else apple_close.diff().dropna()
msft_series = msft_close if msft_stationary else msft_close.diff().dropna()

# Fit ARMA models
def fit_arma(series, name):
    try:
        order_selection = arma_order_select_ic(series, ic='bic', trend='c')
        p, q = order_selection.bic_min_order
        print(f'\nOptimal ARMA order for {name}: AR({p}), MA({q})')
        model = ARIMA(series, order=(p, 0, q))
        results = model.fit()
        print(f'{name} ARMA({p},{q}) Summary:')
        print(results.summary())
    except Exception as e:
        print(f'Error fitting ARMA for {name}: {e}')

fit_arma(apple_series, 'Apple')
fit_arma(msft_series, 'Microsoft')

# Johansen cointegration test
def johansen_coint_test(series1, series2):
    data = pd.DataFrame({'Apple': series1, 'Microsoft': series2}).dropna()
    result = coint_johansen(data, det_order=0, k_ar_diff=1)
    print('\nJohansen Cointegration Test Results:')
    print('Trace Statistic:', result.lr1)
    print('Critical Values (90%, 95%, 99%):')
    print(result.cvt)
    for i, stat in enumerate(result.lr1):
        if stat > result.cvt[i, 1]:
            print(f'Cointegration found at rank {i} (95% confidence)')
        else:
            print(f'No cointegration at rank {i} (95% confidence)')

johansen_coint_test(apple_close, msft_close)
```

```python
# Import required libraries
import pandas as pd
import numpy as np
from google.colab import files
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller, kpss
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.vector_ar.vecm import coint_johansen
import matplotlib.pyplot as plt

# Upload Excel files
print("Please upload APPLE 1 MIN Excel file")
apple_upload = files.upload()
apple_filename = list(apple_upload.keys())[0]
apple_df = pd.read_excel(apple_filename)

print("Please upload MICROSOFT 1 MIN Excel file")
msft_upload = files.upload()
msft_filename = list(msft_upload.keys())[0]
msft_df = pd.read_excel(msft_filename)

# Print available columns
print("\nAPPLE 1 MIN DataFrame columns:", list(apple_df.columns))
print("MICROSOFT 1 MIN DataFrame columns:", list(msft_df.columns))

# Prompt user for the correct column name
close_column = input("Please enter the column name containing closing prices (case-sensitive): ")

# Check if the specified column exists and extract closing prices
if close_column not in apple_df.columns or close_column not in msft_df.columns:
    print(f"Error: Column '{close_column}' not found in one or both DataFrames")
    print("Please check the column names above and ensure both files have the specified column")
    raise KeyError(f"Column '{close_column}' not found")

apple_close = apple_df[close_column]
msft_close = msft_df[close_column]

# Ensure both series have the same length
min_length = min(len(apple_close), len(msft_close))
apple_close = apple_close[:min_length]
msft_close = msft_close[:min_length]

# Function for unit root tests with interpretation
def unit_root_tests(series, name):
    print(f"\nUnit Root Tests for {name}:")

    # ADF Test
    adf_result = adfuller(series)
    print("ADF Test:")
    print(f'ADF Statistic: {adf_result[0]:.4f}')
    print(f'p-value: {adf_result[1]:.4f}')
    print(f'Critical Values: {adf_result[4]}')
    print("Interpretation:")
    if adf_result[1] < 0.05:
        print(f"  - p-value < 0.05: Reject null hypothesis - {name} is stationary")
    else:
        print(f"  - p-value >= 0.05: Fail to reject null - {name} may be non-stationary")

    # KPSS Test
    kpss_result = kpss(series, regression='c')
    print("\nKPSS Test:")
    print(f'KPSS Statistic: {kpss_result[0]:.4f}')
    print(f'p-value: {kpss_result[1]:.4f}')
    print(f'Critical Values: {kpss_result[3]}')
    print("Interpretation:")
    if kpss_result[1] < 0.05:
        print(f"  - p-value < 0.05: Reject null hypothesis - {name} is non-stationary")
    else:
        print(f"  - p-value >= 0.05: Fail to reject null - {name} may be stationary")

# Perform unit root tests
unit_root_tests(apple_close, "APPLE")
unit_root_tests(msft_close, "MICROSOFT")

# Cointegration test with interpretation
def cointegration_test(df):
    result = coint_johansen(df, det_order=0, k_ar_diff=1)
    print("\nJohansen Cointegration Test:")
    print(f"Trace statistic: {result.lr1}")
    print(f"Critical values (90%, 95%, 99%): {result.cvt}")
    print("Interpretation:")
    for i in range(len(result.lr1)):
        if result.lr1[i] > result.cvt[i, 1]:
            print(f"  - r = {i}: Cointegration exists at 95% confidence level")
            print(f"    Trace statistic ({result.lr1[i]:.2f}) > 95% critical value ({result.cvt[i, 1]:.2f})")
        else:
            print(f"  - r = {i}: No cointegration at 95% confidence level")
            print(f"    Trace statistic ({result.lr1[i]:.2f}) <= 95% critical value ({result.cvt[i, 1]:.2f})")
    if result.lr1[0] > result.cvt[0, 1]:
        print("Conclusion: APPLE and MICROSOFT are cointegrated - they share a long-run equilibrium relationship")
    else:
        print("Conclusion: No evidence of cointegration between APPLE and MICROSOFT")

# Prepare data for cointegration
coint_df = pd.DataFrame({
    'APPLE': apple_close,
    'MICROSOFT': msft_close
}).dropna()
cointegration_test(coint_df)

# Function to find best ARIMA model with interpretation
def find_best_arima(series, name, max_p=3, max_d=2, max_q=3):
    best_aic = float('inf')
    best_order = None

    for p in range(max_p + 1):
        for d in range(max_d + 1):
            for q in range(max_q + 1):
                try:
                    model = ARIMA(series, order=(p, d, q))
                    results = model.fit()
                    if results.aic < best_aic:
                        best_aic = results.aic
                        best_order = (p, d, q)
                except:
                    continue

    print(f"\nBest ARIMA model for {name}:")
    print(f"Order: {best_order}")
    print(f"AIC: {best_aic:.2f}")
    print("Interpretation:")
    print(f"  - p={best_order[0]}: {best_order[0]} autoregressive term(s)")
    print(f"  - d={best_order[1]}: {best_order[1]} difference(s) needed for stationarity")
    print(f"  - q={best_order[2]}: {best_order[2]} moving average term(s)")
    return best_order

# Find and fit best ARIMA models
apple_order = find_best_arima(apple_close, "APPLE")
msft_order = find_best_arima(msft_close, "MICROSOFT")

# Fit final ARIMA models
apple_model = ARIMA(apple_close, order=apple_order).fit()
msft_model = ARIMA(msft_close, order=msft_order).fit()

# Forecast next 30 periods (30 minutes for 1-min data)
forecast_steps = 30
apple_forecast = apple_model.forecast(steps=forecast_steps)
msft_forecast = msft_model.forecast(steps=forecast_steps)

# Create forecast index
last_index = len(apple_close) - 1
forecast_index = range(last_index + 1, last_index + 1 + forecast_steps)

# Plot original series with forecasts
plt.figure(figsize=(12,6))
plt.plot(apple_close, label='APPLE Historical')
plt.plot(forecast_index, apple_forecast, label='APPLE Forecast', color='red')
plt.plot(msft_close, label='MICROSOFT Historical')
plt.plot(forecast_index, msft_forecast, label='MICROSOFT Forecast', color='green')
plt.title('APPLE and MICROSOFT Closing Prices with Forecasts (1-Min)')
plt.legend()
plt.show()

# Detailed forecast plot with confidence intervals and interpretation
def plot_forecast(model, series, name, steps=30):
    forecast_obj = model.get_forecast(steps=steps)
    forecast = forecast_obj.predicted_mean
    conf_int = forecast_obj.conf_int()

    forecast_index = range(len(series), len(series) + steps)

    plt.figure(figsize=(12,6))
    plt.plot(series, label=f'{name} Historical')
    plt.plot(forecast_index, forecast, label='Forecast', color='red')
    plt.fill_between(forecast_index,
                    conf_int.iloc[:, 0],
                    conf_int.iloc[:, 1],
                    color='pink',
                    alpha=0.3,
                    label='95% Confidence Interval')
    plt.title(f'{name} Price Forecast (1-Min)')
    plt.legend()
    plt.show()

    # Forecast interpretation
    last_value = series.iloc[-1]
    mean_forecast = forecast.mean()
    print(f"\nForecast Interpretation for {name}:")
    print(f"Last observed value: {last_value:.2f}")
    print(f"Average forecast value: {mean_forecast:.2f}")
    print(f"Forecast change: {mean_forecast - last_value:.2f}")
    if mean_forecast > last_value:
        print("Trend: Upward forecast trend")
    elif mean_forecast < last_value:
        print("Trend: Downward forecast trend")
    else:
        print("Trend: Flat forecast trend")
    print(f"95% CI range at period {steps}: [{conf_int.iloc[-1, 0]:.2f}, {conf_int.iloc[-1, 1]:.2f}]")

# Generate detailed forecast plots and interpretations
plot_forecast(apple_model, apple_close, "APPLE")
plot_forecast(msft_model, msft_close, "MICROSOFT")

# Print forecast values
print("\nAPPLE Forecast Values (next 5 periods):")
print(apple_forecast[:5])
print("\nMICROSOFT Forecast Values (next 5 periods):")
print(msft_forecast[:5])
```
